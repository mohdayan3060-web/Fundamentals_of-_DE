Basic Date and time in python 


```python
# date module 
#  (YYYY-MM-DD ) -> format
from datetime import date
date.today()

date.today().year
date.today().month
date.today().day


```

```python
#time module 
from datetime import time
meeting_time =time(14,30,10,50000)
meeting_time

#extarcting part 
meeting_time.hour
meeting_time.minute
meeting_time.second
meeting_time.microsecond
```

```python
#date and time combo 
from datetime import datetime
datetime.now()

#extarcting part 
datetime.now().hour
datetime.now().year
datetime.now().time()
```

```python
#timeDelta 
# A timedelta object represents a duration or the difference between two dates/times.

from datetime import datetime ,timedelta
now= datetime.now()

#create duration 
five_days= timedelta(days=5)
sevenHours=timedelta(hours=7)
half_hours=timedelta(minutes=30)

#addition and Substraction 
five_days_form_now= now+five_days

seven_hours_ago=now-sevenHours

print(five_days_form_now)
print(seven_hours_ago)
```

```python
# Parse Date Strings (String ➡️ Datetime)
# Convert String into Date and time Object 

from datetime import datetime
date_str= '25-10-2024 14:30'

parse_datetime=datetime.strptime(date_str,"%d-%m-%Y %H:%M")
parse_datetime

#extarcting 
parse_datetime.day
parse_datetime.month
parse_datetime.hour
parse_datetime.year
type(parse_datetime)
```

```python
#Formate dates(Datetime  -->> String )
# It convert Readable String formate 

from datetime import datetime

now= datetime.now()

print(now.strftime('%Y-%d-%m'))
print(now.strftime('%Y-%b-%d')) # %b -> for sort months names 
print(now.strftime('%I:%M %p')) #%I -> fro 12-hours clock  and %p -> for AM/PM
print(now.strftime('Today is %A')) #%A is full week name 

```

```python
# In Python, a datetime object is naive if it does not contain any timezone information. Technically, its tzinfo (timezone info) attribute is set to None.
# By default, almost all datetime objects you create in Python are naive.

from datetime import datetime
now =datetime(2026, 9, 22)
print(now.tzinfo) #None  means no timezone Idea 
```
