```python
import numpy as np 
import pandas as pd
#Pandas series -> 1-d array which store data of datatype , example -> each coulmn is series 
# in pandas  string datatype is object 
```

```python
#create series from list of string type 
country= ['India','USA','Nepal' ,'China']
pd.Series(country)
```

```python
#integer series 
Run=[100,34,56,78,98,90]
pd.Series(Run)
```

```python
#custom index 
 
marks=[90,87,88,67,56]
subject=['cs','java','oops','math','DBMS']

marks_series=pd.Series(marks,index=subject)

#inserting name
pd.Series(marks ,name='StudentMarks')
```

```python
# Read csv
# if you have multiple columns  then use this snipt for converting in  pandas series 

customer_data=pd.read_csv('customers.csv', index_col='name', usecols=['name','age'])['age']
customer_data
```

Series Methods 

```python
# head(): Returns the first few items (default 5) of the Series 
customer_data.head(10)
# tail(): Returns the last few items of the Series 
customer_data.tail(5)
# sample(): Returns a random selection of items from the Series 
# Sorting and Analysis:
customer_data.sample(10)

# value_counts(): Returns the frequency of each unique value in the Series 
customer_data.value_counts()
# sort_values(): Sorts the Series based on its values 

customer_data.sort_values(ascending=True) #use inplace= True  (for Prmanent chnages )

# sort_index(): Sorts the Series based on its index labels 
customer_data.sort_index(ascending=False)
```

Pandas series mathmatical induction 

```python
# count(): Returns the number of non-missing values in the Series 
customer_data.count()
# sum(): Calculates the total sum of all values in the Series 
customer_data.sum()
# product(): Multiplies all values in the Series together (1:07:00).
customer_data.product()
# mean(): Computes the arithmetic average of the values (1:08:35).
customer_data.mean()
# median(): Finds the middle value of the data (1:08:35).
customer_data.median()
# mode(): Identifies the most frequently occurring value 
customer_data.mode()
# std(): Calculates the standard deviation of the data
customer_data.std()
# var(): Computes the variance of the data 
customer_data.var()
# min(): Returns the smallest value in the Series 
customer_data.min()
# max(): Returns the largest value in the Series 
customer_data.max()
# describe(): Provides a comprehensive summary including count, mean, standard deviation, min, max, and percentiles
customer_data.describe()
```
