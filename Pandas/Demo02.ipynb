Aggregates with Pandas 

```python
order['sales'].sum()
order['sales'].min()
order['sales'].max()
order['sales'].mean()
order['sales'].median()
order['region'].unique()
order['category'].value_counts()

```

```python
#Group by 
order.groupby('region')['sales'].sum()
order.groupby('order_id')['sales'].count().max()
order.groupby(['region','category'])['quantity'].agg(['sum','max','min'])
order.groupby(['order_date','category'])['profit'].sum().sort_values(ascending=False)

#aggreagtion on multiple columns 
order.groupby('category').agg({'sales':sum, 'profit':'max'})
order.groupby(['region','category']).agg({'quantity':'sum','profit':'sum'})
```

```python
#pivot table
pd.pivot_table(
    order,
    index='region',
    columns='category',
    values='profit',
    aggfunc='max',
    fill_value=0

)

pd.pivot_table(
    order,
    index='product_id',
    columns='category',
    values='quantity',
    aggfunc='max',
    fill_value=0


).sample(3)
```
