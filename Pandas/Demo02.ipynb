```python
#dealing with null Values and  droping dublicates Values 
customer=pd.read_csv('customers.csv')


# #check na_value 
customer.isnull().sum()
# #droping null values 

customer.dropna(subset='age', inplace=True )

customer.info()
```

```python
#filling null Values  |Specific Columns 
customer.isnull().sum()
customer['state']=customer['state'].fillna('unknown')
# customer.fillna({'state': 'Unknown'}, inplace=True)

customer.isnull().sum()

```

```python
#droping dublicates 
customer[customer.duplicated()] 

customer.drop_duplicates(keep='last',inplace=True)
customer.duplicated().sum()
```
