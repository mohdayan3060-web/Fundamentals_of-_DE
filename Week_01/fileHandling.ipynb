```python
#Serialization and Deserialization 

#serialization  means  converting python data type into JSON formate 
#Deserialization means converting JSON to python data type 
```

```python
#list 
import json
l=[10,20,30,40,50,60,70,80]
with open('demo.json', 'w') as f:
    json.dump(l,f,indent=4)

#deserialize 
with open('demo.json','r') as f:
    print(type(json.load(f))) 
```

```python
# dict 
import json 
d={
    "name ":"ayan",
    "age ":21,
    "collage":"IUL"
}

with open('demo.json','w') as f:
    json.dump(d,f,indent=3)

#reading json file
with open('demo.json','r') as f:
    dicto=json.load(f)
    print(dicto)
    print(type(dicto))
   
```

```python
# Tuple 
tuple =(10,20,30,"name")
with open('demo.json' ,'w') as f:
    json.dump(tuple,f)

with open('demo.json', 'r')as f:
    print(type(json.load(f)))
```

Here is why tuples become lists in JSON:JSON doesn't know what a tuple is: The JSON format only recognizes a few basic data types, and "tuple" is not one of them.Tuples look like arrays: To make it work, Python converts your tuple into a JSON array (which looks like [1, 2, 3]).Arrays always load as lists: When Python reads that JSON array back, it automatically converts it into a Python list because it cannot tell it used to be a tuple.Information is lost: The distinction between a tuple and a list disappears the moment the data is saved as JSON text.

Pickling in Python is the process of converting a Python object (like a list, dictionary, or custom class) into a byte stream so it can be saved to a file or sent over a network.When you want to load that file back into Python, you unpickle it to restore the exact original object.Unlike JSON, pickling preserves Python-specific data types, including tuples.
