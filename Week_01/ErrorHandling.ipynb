```python
#Handling exception in python 
#maintance flow 


try:
    n=10
    d=int(input("Enter Denominator "))
    ans=n/d
    print(ans)
    #code that might be trow error 
    
except:
    print("Error occured ")
    #catching error  when try block throw error 
  
else :
    #else block get executed when try block does not raise any error 
    #only execute when no exception is raise 
    print("this is else block")
    try:
        print(1/0)
    except Exception as e:
        print(e)    
    pass
finally:
    # Always  run does not matter Exception is raised or not 
    #it is often used for cleanup and  action like ('Closing file ') and reslasing resources 
    print("finally block always execute ")
    print('closing all file ')
    # it execute  even you return function 

```

```python

#Raise 
#python allow to raise forcully error 

try: 
    age=int(input("Enter your age "))
    if age <18 :
        raise ValueError ('make sure age above 18')
except ValueError as e:
    print(e)
except TypeError as e:
    print(e)
except Exception as e:
    print(e)
        
```

```python
#custom error  
#python allow to create Custom exception by creating class 

class WrongEmail(Exception):
    def __init__(self,msg):
        super().__init__(msg)

try:
    email=input("Enter your email")
    if '@' not in email:
        raise WrongEmail("check your eamil again")
    
except WrongEmail as e:
    print("Enter your email again")
    print(e)
```
