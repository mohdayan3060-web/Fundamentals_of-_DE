List is a built-in data structure used to store an ordered collection of items. They are dynamic, resizable and capable of storing multiple data types.
Mutable: list elements can be changed, updated, added, or removed after the list is created.
Ordered: elements maintain the order in which they are inserted.
Index-based: elements are accessed using their position, starting from index 0.

```python
my_list =[10,20, 20,30.5,True,"ayan",None, 3+1j] #repeted elements , differnt data types , 
my_list
#loop on list 
for item in my_list:
    print(item)
```

```python
# Adding elements
new_list=[20,33,43]
new_list.append(100)  # Adds an element at the end of the list.
new_list
new_list.insert(2,101) #insert(inde, elemtns )
new_list.insert(0,"hey!")
new_list
# extend(): Adds multiple elements to the end of the list.
new_list.extend(["ayan", True, None ])
new_list
#remove (elements )
new_list.remove("ayan")
new_list
new_list.pop() #Removes the element at a specific index or the last element if no index is specified.
new_list

del new_list[3]
new_list

new_list.clear()
new_list
```
