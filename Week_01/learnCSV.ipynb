```python
#Converters 
def rename(name):
    if name=='Royal Challengers Bangalore':
        return 'RCB'
    elif name=='Kolkata Knight Riders':
        return 'KKR'
    else:
        return name
pd.read_csv('Match_Info.csv' , converters={'team1':rename, 'team2':rename , 'toss_winner':rename})
```
