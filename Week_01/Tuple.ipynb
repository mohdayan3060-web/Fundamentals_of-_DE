A tuple is an immutable ordered collection of elements.

Tuples are similar to lists, but unlike lists, they cannot be changed after their creation.
Can hold elements of different data types.
These are ordered, heterogeneous and immutable.

```python
my_tuple=( 5, "wellcome ", True ,None, 3.2, {'key' :'value '}, )
my_tuple
type(my_tuple)
my_tuple
for item in my_tuple:
    print(item)
my_tuple[0]     # 5

```

```python
tup1=(1,2,3)
tup2=("ayan", "rahul","faiz")
tup3=tup1+tup2 #Concatination 
tup3
```

```python
# The slicing syntax is tuple[start:stop:step]
tup4=("microscopic")
tup4[::-1] #for reverse
tup4[2:6:1]
```
