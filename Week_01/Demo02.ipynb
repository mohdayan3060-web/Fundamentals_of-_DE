The core difference is that mutable objects can be changed after they are created, while immutable objects cannot.
🧱 Immutable Objects (Unchangeable)Think of these as objects locked in stone
.Once made, you cannot edit them.Examples: Integers (int), text (str), and tuples (tuple).
What happens during updates: If you change a text string, Python copies the old text, makes the change, and saves it as a brand new item.
The Tuple Exception: A tuple is locked. However, if you put a changeable list inside a tuple, you can still edit that specific list.

📝 Mutable Objects (Changeable)Think of these as objects written on a whiteboard.
 You can erase, add, or rewrite parts of them freely.Examples: 
Lists (list), dictionaries (dict), and sets (set).What happens during updates:
Python edits the data exactly where it is. The object keeps its original identity.

 Identity Operators (is / is not)These operators check if two variables point to the exact same spot in computer memory.
 Memebership Operators check if the content exists inside the group. we use ( in ).
 

```python
# Two separate grocery carts with the same items
my_cart = ["apples", "milk"]
your_cart = ["apples", "milk"]

# Do they contain the same things?
print(my_cart == your_cart)  # Output: True  (The values are equal)

# Are they the exact same physical cart?
print(my_cart is your_cart)  # Output: False (They are separate objects in memory)
```
