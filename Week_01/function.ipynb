```python
#keywords Arguments 
#  pass values using parameter names, so argument order does not matter.


def student(fname, lname):
    print(fname,lname)

student(fname="Ayan" ,lname="Ansari ")
student(lname="Anasri ", fname="Ayan")


```

```python
# 3. Positional Arguments: values are assigned to parameters based on their order in the function call.
def student(name,id ):
    print(f"Name :{name} and Id :{id}")

student("ayan",2500101563)    
student(2500101563, "ayan")
```

```python

#  Arbitrary Arguments: allow functions to accept multiple values. This is done using two special symbols:
# *args -> Variable length Arguments | treated as tuple | non -keywords arguments |

def marks(*args):
    print(args)


marks(90,100,45,67,87,98)


```

```python
# **kargs -> Variable length  keywords arguments , |Treated as dictionary 

def greet(**kargs):
    print(kargs)

    for key , val in kargs.items():
        print(key,val)

greet(First='Hello' , second='Good Morning')

```

```python
# Parameter vs arguments 
def function(x,y): #x , y are prameters 
    return x+y

function (10,20) #10, 20 is Arguments 
# a parameter is a variable in a function or method declaration. It serves as a placeholder for data that will be provided when the function or method is called. Parameters define the type and number of values that a function or method can accept.

# Arguments, also known as actual arguments, are the values supplied to a function when it is called. These values serve as inputs to the function during its execution.
```
